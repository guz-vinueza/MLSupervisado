```python
# ==========================================
# Modelo Supervisado con KNN
# Dataset público: Wine Dataset
# ==========================================

# 1. Importar librerías
import pandas as pd

from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
```

```python
# 2. Cargar el dataset
wine = load_wine()

X = wine.data          # Variables predictoras
y = wine.target        # Variable objetivo: tipo de vino
```

El dataset de vino contiene propiedades químicas de diferentes vinos, como el alcohol, el ácido málico, la ceniza, etc.
La etiqueta (variable `y`) representa el tipo o clase de vino, indicando su origen o variedad.
Permite clasificar los vinos basándose en sus características analíticas.

```python
# 3. Convertir a DataFrame solo para visualizar mejor
df = pd.DataFrame(X, columns=wine.feature_names)
df["target"] = y

print("Primeras filas del dataset:")
print(df.head())
```

```python
# 4. Dividir los datos en entrenamiento y prueba
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42
)
```

```python
# 5. Estandarizar los datos
# KNN usa distancias, por eso es importante que las variables estén en la misma escala
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)   # Aprende la escala con datos de entrenamiento
X_test_scaled = scaler.transform(X_test)         # Aplica la misma escala a los datos de prueba
```

```python
# 6. Crear el modelo KNN, con k=5
model = KNeighborsClassifier(n_neighbors=5)
```

```python
# 7. Entrenar el modelo
model.fit(X_train_scaled, y_train)
```

```python
# 8. Hacer predicciones
y_pred = model.predict(X_test_scaled)
```

```python
# 9. Evaluar el modelo
accuracy = accuracy_score(y_test, y_pred)

print("\nAccuracy del modelo:")
print(accuracy)

print("\nMatriz de confusión:")
print(confusion_matrix(y_test, y_pred))

print("\nReporte de clasificación:")
print(classification_report(y_test, y_pred))
```

El modelo KNN obtuvo una **precisión general del 94.4%**, lo que indica que clasificó correctamente la mayoría de los vinos.
La **matriz de confusión** muestra un rendimiento excelente para las clases 0 y 2, con algunas confusiones mínimas entre la clase 1 y la clase 2.
El **reporte de clasificación** detalla que el modelo es robusto para todas las clases, con alta precisión y recall, especialmente para las clases 0 y 2.

```python
# 10. Usar el modelo con un nuevo caso
nuevo_vino = [X_test[5]]  # Tomamos un ejemplo del set de prueba

# Escalamos la data
nuevo_vino_scaled = scaler.transform(nuevo_vino)

prediccion = model.predict(nuevo_vino_scaled)

print("\nPredicción para un nuevo vino:")
print("Clase predicha:", prediccion[0])
```
