```
min_year = df['year'].min()
max_year = df['year'].max()

print(f"Minimum Year: {min_year}")
print(f"Maximum Year: {max_year}")
```

```
driver_participations = df['Driver'].value_counts().reset_index().head(10)
driver_participations.columns = ['Driver', 'Number of Participations']

print(driver_participations.to_markdown(index=False, numalign="left", stralign="left"))
```

```
import matplotlib.pyplot as plt
import seaborn as sns

car_participations_top10 = df['Car'].value_counts().reset_index().head(10)
car_participations_top10.columns = ['Car', 'Number of Participations']

plt.figure(figsize=(12, 7))
sns.barplot(x='Number of Participations', y='Car', data=car_participations_top10, palette='viridis', hue='Car', legend=False)
plt.title('Top 10 Cars by Number of Participations')
plt.xlabel('Number of Participations')
plt.ylabel('Car')
plt.grid(axis='x', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```
df['Pos'] = pd.to_numeric(df['Pos'], errors='coerce')
top_3_finishes = df[df['Pos'].isin([1, 2, 3])]
nationality_top_3_counts = top_3_finishes['Nationality'].value_counts().reset_index().head(10)
nationality_top_3_counts.columns = ['Nationality', 'Top 3 Finishes']

print(nationality_top_3_counts.to_markdown(index=False, numalign="left", stralign="left"))
```
