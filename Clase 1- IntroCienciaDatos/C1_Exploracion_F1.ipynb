### Ejemplos Gráficos

```
# Sección de Librerías
import matplotlib.pyplot as plt
import seaborn as sns

# Obtiene la tabla de autos y número de participaciones
car_participations_top10 = df['Car'].value_counts().reset_index().head(10)
car_participations_top10.columns = ['Escudería', '# Participaciones']
car_participations_top10
```

```
# Prepara el gráfico.  Le da tamaño
plt.figure(figsize=(12, 7))
# Gráfico de barras
sns.barplot(x='# Participaciones', y='Escudería', data=car_participations_top10, palette='viridis', hue='Escudería', legend=False)
plt.title('Top 10 Escuderías por # Participaciones')
plt.xlabel('# Participaciones')
plt.ylabel('Escudería')
plt.grid(axis='x', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```
# Convierte el campo Pos a numérico
df['Pos'] = pd.to_numeric(df['Pos'], errors='coerce')
# Arma un dataset con los registros cuando la posición es 1, 2, 3
top_3_finishes = df[df['Pos'].isin([1, 2, 3])]
top_3_finishes

```

```
# Genera una tabla con las nacionalidades y su número de posiciones en podio
# Top 10
nationality_top_3_counts = top_3_finishes['Nationality'].value_counts().reset_index().head(10)
nationality_top_3_counts.columns = ['Nationality', 'Top 3 Finishes']
nationality_top_3_counts
```

```
# Top 10 de Pilotos con Carreras Ganadas
races_won = df[df['Pos'] == 1]
races_won
```

```
# Agrupa a los ganadores según el piloto (Driver)
# Genera un agregado de el número de victorias, el mínimo y máximo año
wins_summary = races_won.groupby('Driver').agg(
    total_wins=('Pos', 'count'),
    min_year=('year', 'min'),
    max_year=('year', 'max')
).reset_index()
wins_summary = wins_summary.sort_values(by='total_wins', ascending=False)
wins_summary

#wins_summary.head(10)

```
